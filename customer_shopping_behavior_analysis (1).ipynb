```python
#Use transform() to calculate category average
df_clean["category_average"] = (df_clean.groupby("category")["purchase_amount_(usd)"].transform("mean"))
df_clean[["category", "purchase_amount_(usd)", "category_average"]].head(10)
```

```python
#Use apply() for custom classification
def clasify_spending(amount):
    if amount < 50:
        return "low"
    elif amount < 80:
        return "medium"
    else:
        return "high"

df_clean["spending_class"] = (df_clean["purchase_amount_(usd)"].apply(clasify_spending))
df_clean[["purchase_amount_(usd)", "spending_class"]]
```

```python
#Use np.where() for discount classification
import numpy as np
df_clean["discount_status"] = np.where(df_clean["discount_applied"] == "Yes", "discount used" , "no discount")
df_clean[["discount_applied", "discount_status"]].head(10)
```

```python
#Use NumPy mean, median and standard deviation
import numpy as np
purchase = df_clean["purchase_amount_(usd)"].to_numpy()
print("Mean:", np.mean(purchase))
print("Median:", np.median(purchase))
print("standard_deviation:", np.std(purchase))
```

```python
#Use NumPy percentiles to analyze spending
purchase = df_clean["purchase_amount_(usd)"].to_numpy()
print("25th Percentile:", np.percentile(purchase, 25))
print("50th Percentile:", np.percentile(purchase, 50))
print("75th Percentile:", np.percentile(purchase, 75))
print("90th Percentile:", np.percentile(purchase, 90))
```

```python
#Detect outliers using IQR
purchase = df_clean["purchase_amount_(usd)"]
Q1 = purchase.quantile(0.25)
Q3 = purchase.quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5*IQR
upper_limit = Q1 + 1.5*IQR
print("lower_limit", lower_limit)
print("upper_limit", upper_limit)
```

```python
outlier = (df_clean["purchase_amount_(usd)"] < lower_limit) | (df_clean["purchase_amount_(usd)"] > upper_limit)
outlier
```

```python
outlier.shape[0]
```

```python
#Use np.select() for customer segmentation
conditions = [(df_clean["purchase_amount_(usd)"] >= 80) & (df_clean["subscription_status"] == "yes"), 
              (df_clean["purchase_amount_(usd)"] >= 80), (df_clean["purchase_amount_(usd)"] < 50)]
choice = ["Premium Subscriber", "High Spender" , "Low Spender"]

df_clean["customer_segment"] = np.select(conditions, choice, default="regular customer")
df_clean[["purchase_amount_(usd)", "subscription_status", "customer_segment"]].head(10)
```

```python
#Compare spending patterns across segments
df_clean.groupby("customer_segment").agg(
    total_revenue=("purchase_amount_(usd)", "sum"),
    average_spending=("purchase_amount_(usd)", "mean"),
    median_spending=("purchase_amount_(usd)", "median"),
    customer_count=("customer_id", "count")
).sort_values("total_revenue", ascending=False)
```
